```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Convert order date to datetime for time-based analysis
df[['Order Date']] = pd.to_datetime(df['Order Date'])

df.head(10)
```

```python
# Sales and profit by category
cat_summary = df.groupby("Category")[["Sales","Profit"]].sum().sort_values("Sales", ascending=False)
print(cat_summary)

# Plot
plt.figure(figsize=(8,5))
sns.barplot(data=cat_summary.reset_index(), x="Category", y="Sales")
plt.title("Sales by Category")
plt.show()

plt.figure(figsize=(8,5))
sns.barplot(data=cat_summary.reset_index(), x="Category", y="Profit")
plt.title("Profit by Category")
plt.show()

```

```python
# Sub-category
subcat_summary = df.groupby("Sub Category")[["Sales","Profit"]].sum().sort_values("Profit", ascending=False)
print(subcat_summary)
plt.figure(figsize=(12,5))
sns.barplot(data=subcat_summary.reset_index(), x="Sub Category", y="Profit")
plt.title("Profit by Sub-Category")
plt.xticks(rotation=90)
plt.show()
```

```python
# Done previously.
# Average profit vs discount
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="Discount", y="Profit")
plt.title("Profit vs Discount")
plt.show()

# Profit grouped by discount levels
df['Discount Level'] = pd.cut(df['Discount'], bins=[0,0.2,0.5,1], labels=["Low","Medium","High"])
print(df.groupby("Discount Level")["Profit"].mean())
```

```python
# Top 10 customers by profit
cust_summary = df.groupby("Customer Name")[["Sales","Profit"]].sum().sort_values("Profit", ascending=False).head(10)
print(cust_summary)

plt.figure(figsize=(10,5))
sns.barplot(data=cust_summary.reset_index(), x="Customer Name", y="Profit")
plt.title("Top 10 Customers by Profit")
plt.xticks(rotation=45)
plt.show()
# (groupby results for plotting), we use reset_index() because we need the group names (Category, Sub Category, etc.) as columns for seaborn plots.
```

```python
loss_subcat = df.groupby("Sub Category")[["Profit"]].sum().sort_values("Profit")
print(loss_subcat.head())

plt.figure(figsize=(10,5))
sns.barplot(data=loss_subcat.reset_index(), x="Sub Category", y="Profit")
plt.title("Profit by Sub-Category (Check Losses)")
plt.xticks(rotation=45)
plt.show()
```

```python
# 1. Overall summary: Sales, Profit, Discount
overall_summary = {
    "Total Sales": df["Sales"].sum(),
    "Total Profit": df["Profit"].sum(),
    "Average Discount": df["Discount"].mean()
}
print(overall_summary)
```

```python
# 2. Category-wise sales & profit
category_summary = df.groupby("Category")[["Sales","Profit"]].sum().sort_values("Sales", ascending=False)
print(category_summary)
```

```python
# 3. Lowest profit-making subcategories
subcat_loss = df.groupby("Sub Category")[["Profit"]].sum().sort_values("Profit").head(5)
print(subcat_loss)
```

```python
# 4. Region-wise sales & profit
region_summary = df.groupby("Region")[["Sales","Profit"]].sum().sort_values("Sales", ascending=False)
print(region_summary)
```

West & East regions dominate sales & profit.

North region negligible → very small presence or new market.

```python
# 5. Discount impact
df['Discount Level'] = pd.cut(df['Discount'], bins=[0,0.2,0.5,1], labels=["Low","Medium","High"])
discount_summary = df.groupby("Discount Level")["Profit"].mean()
print(discount_summary)
```

Profit almost same for low and medium discount levels.
 Discounts don’t hurt margins much --probably applied on high-margin products.

```python
# 6. Monthly sales and profit trend
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['YearMonth'] = df['Order Date'].dt.to_period('M').dt.to_timestamp()
monthly_summary = df.groupby("YearMonth")[["Sales","Profit"]].sum()
print(monthly_summary.head(10))  # show first 10 months
```

Sales & profit fluctuate each month.

Example: March & April show peaks, while February is weak.
Indicates seasonality in demand.

```python
import matplotlib.pyplot as plt

# Total Sales vs Profit
values = [df["Sales"].sum(), df["Profit"].sum()]
labels = ["Sales","Profit"]

plt.figure(figsize=(5,5))
plt.pie(values, labels=labels, autopct="%1.1f%%", startangle=90)
plt.title("Sales vs Profit Contribution")
plt.show()
```

Profit is a smaller slice of sales → business runs on thin margins.

```python
import seaborn as sns

category_summary = df.groupby("Category")[["Sales","Profit"]].sum().reset_index()

plt.figure(figsize=(10,5))
sns.barplot(data=category_summary, x="Category", y="Sales")
plt.title("Sales by Category")
plt.show()

plt.figure(figsize=(10,5))
sns.barplot(data=category_summary, x="Category", y="Profit")
plt.title("Profit by Category")
plt.show()
```

Eggs, Meat & Fish + Snacks lead in both sales & profit.

Categories are balanced overall.

```python
subcat_loss = df.groupby("Sub Category")["Profit"].sum().sort_values().head(10).reset_index()

plt.figure(figsize=(10,6))
sns.barplot(data=subcat_loss, x="Profit", y="Sub Category", palette="Reds_r")
plt.title("Lowest Profit Subcategories")
plt.show()
```

Staples (Rice, Pulses, Flour, Organic Fruits) have low profit.
 Attract customers but don’t drive margins.

```python
region_summary = df.groupby("Region")[["Sales","Profit"]].sum().reset_index()

fig, axes = plt.subplots(1,2, figsize=(14,5))

sns.barplot(data=region_summary, x="Region", y="Sales", ax=axes[0])
axes[0].set_title("Sales by Region")

sns.barplot(data=region_summary, x="Region", y="Profit", ax=axes[1])
axes[1].set_title("Profit by Region")

plt.show()
```

West & East regions are strong markets.

North negligible.

```python
df['Discount Level'] = pd.cut(df['Discount'], bins=[0,0.2,0.5,1], labels=["Low","Medium","High"])

discount_summary = df.groupby("Discount Level")["Profit"].mean().reset_index()

plt.figure(figsize=(6,4))
sns.barplot(data=discount_summary, x="Discount Level", y="Profit", palette="coolwarm")
plt.title("Average Profit per Order vs Discount Level")
plt.show()
```

Profit/order almost same for Low & Medium discounts.

Discounts not reducing margins drastically.

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['YearMonth'] = df['Order Date'].dt.to_period('M').dt.to_timestamp()

monthly_summary = df.groupby("YearMonth")[["Sales","Profit"]].sum().reset_index()

plt.figure(figsize=(12,6))
sns.lineplot(data=monthly_summary, x="YearMonth", y="Sales", label="Sales")
sns.lineplot(data=monthly_summary, x="YearMonth", y="Profit", label="Profit")
plt.title("Monthly Sales & Profit Trend")
plt.xticks(rotation=45)
plt.legend()
plt.show()
```
